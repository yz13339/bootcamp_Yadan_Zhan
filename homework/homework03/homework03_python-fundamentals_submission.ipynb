# Homework 03: Python Fundamentals

This notebook demonstrates NumPy operations, pandas data inspection,
summary statistics, groupby aggregation, and reusable functions.

```python
import numpy as np
import pandas as pd
```

```python
## Numpy operations
values = np.array([1, 2, 3, 4, 5])

print("Original array:", values)
print("Add 10:", values + 10)
print("Multiply by 2:", values * 2)
print("Squared:", values ** 2)
```

```python
## read csv
df = pd.read_csv("data/raw/starter_data.csv")

df.head()
```

```python
df.info()
```

```python
print(df.columns.tolist())
```

```python
## summary statistics
summary = df.describe()

summary
```

```python
## reusable function
def get_summary_stats(data):
    return data.describe()

summary = get_summary_stats(df)

summary
```

```python
## groupby
grouped_summary = df.groupby("category").mean(
    numeric_only=True
)

grouped_summary
```

```python
from pathlib import Path

processed_dir = Path("data/processed")
processed_dir.mkdir(parents=True, exist_ok=True)

summary_file = processed_dir / "summary.csv"
summary.to_csv(summary_file)

print("Summary saved to:", summary_file)
print("File exists:", summary_file.exists())
```
